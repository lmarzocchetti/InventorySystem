```python
loss_function_vals = None
with open('loss_function_values.pkl', 'rb') as f:
    loss_function_vals = pickle.load(f)
```

```python
float_list = [t.item() for t in loss_function_vals]
```

```python
plt.plot(float_list)
plt.xlabel('Time-Step')
plt.ylabel('Loss Function')
plt.title('Loss Function Values')
plt.show()
```
